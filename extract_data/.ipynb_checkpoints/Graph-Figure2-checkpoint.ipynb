```python
# Display basic info about the dataset
print("Dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst few rows:")
print(df.head())

# Clean and prepare the data
df['year'] = pd.to_numeric(df['year'], errors='coerce')
df = df.dropna(subset=['year'])
df['year'] = df['year'].astype(int)

# Ensure 'winner' column is boolean (True/False)
df['is_winner'] = df['winner'].astype(bool)

# Aggregate data by year
yearly_stats = df.groupby('year').agg({
    'project_name': 'count',
    'is_winner': 'sum',
    'hackathon_name': 'nunique'
}).rename(columns={
    'project_name': 'total_projects',
    'is_winner': 'total_winners',
    'hackathon_name': 'unique_hackathons'
})

print("\nYearly Statistics:")
print(yearly_stats)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Set IEEE conference style parameters
plt.rcParams.update({
    'font.size': 10,
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'axes.linewidth': 0.8,
    'axes.spines.left': True,
    'axes.spines.bottom': True,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'xtick.major.size': 4,
    'xtick.minor.size': 2,
    'ytick.major.size': 4,
    'ytick.minor.size': 2,
    'legend.frameon': False,
    'figure.dpi': 300
})

# Create the figure with IEEE standard size (double column: 7")
fig, ax = plt.subplots(figsize=(6, 3))

# Calculate non-winners (participants who didn't win)
non_winners = yearly_stats['total_projects'] - yearly_stats['total_winners']

# Create stacked bar chart
width = 0.6  # Width of bars
years = yearly_stats.index

# Plot stacked bars
bars1 = ax.bar(years, non_winners, width, 
               label='Non-Winners', color='#2E86AB', alpha=0.8,
               edgecolor='white', linewidth=0.8)

bars2 = ax.bar(years, yearly_stats['total_winners'], width, 
               bottom=non_winners, label='Winners', color='#A23B72', alpha=0.8,
               edgecolor='white', linewidth=0.8)

# Add value labels on each segment
for i, year in enumerate(years):
    # Label for non-winners (middle of bottom segment)
    non_winner_height = non_winners.iloc[i]
    ax.text(year, non_winner_height/2, 
            f'{non_winner_height}', 
            ha='center', va='center', fontsize=8, 
            color='white', fontweight='bold')
    
    # Label for winners (middle of top segment)
    winner_height = yearly_stats['total_winners'].iloc[i]
    ax.text(year, non_winner_height + winner_height/2, 
            f'{winner_height}', 
            ha='center', va='center', fontsize=8, 
            color='white', fontweight='bold')
    
    # Total value at the top of each bar
    total_height = yearly_stats['total_projects'].iloc[i]
    ax.text(year, total_height + 20, 
            f'{total_height}', 
            ha='center', va='bottom', fontsize=8, 
            color='black', fontweight='bold')

# Title and labels
ax.set_title("Figure 2: Hackathon Participation and Success Rates Over Time", 
             fontsize=10, fontweight='bold', pad=15)
ax.set_xlabel("Year", fontsize=10, fontweight='bold')
ax.set_ylabel("Number of Participants", fontsize=10, fontweight='bold')

# Format x-axis
ax.set_xticks(years)
ax.set_xticklabels(years)

# Set y-axis limits with some padding
max_height = yearly_stats['total_projects'].max()
ax.set_ylim(0, max_height * 1.1)

# Tick styling
ax.tick_params(axis='both', which='major', labelsize=10, direction='in')
ax.tick_params(axis='both', which='minor', direction='in')
ax.minorticks_on()

# Grid styling (only horizontal for bar charts)
ax.grid(True, linestyle='--', alpha=0.4, linewidth=0.5, which='major', axis='y')
ax.grid(True, linestyle=':', alpha=0.2, linewidth=0.3, which='minor', axis='y')

handles, labels = ax.get_legend_handles_labels()
ax.legend(
    handles[::-1], labels[::-1],  # Winners first
    loc='upper left', bbox_to_anchor=(0.956, 1),  # Push legend outside
    fontsize=8,
    framealpha=0, edgecolor='none',
    handlelength=2.5, handletextpad=0.5
)

# Adjust layout
plt.tight_layout(pad=1.2)

# Save the figure at 300 DPI
plt.savefig('hackathon_stacked_bar.png', dpi=300, bbox_inches='tight', 
            facecolor='white', edgecolor='none')
plt.show()
```

```python
# Calculate winner percentages by year
winner_stats = df.groupby('year').agg({
    'is_winner': ['count', 'sum']
}).round(2)

# Flatten column names
winner_stats.columns = ['total_participants', 'total_winners']

# Calculate percentage of winners
winner_stats['winner_percentage'] = (winner_stats['total_winners'] / winner_stats['total_participants'] * 100).round(2)

# Add non-winners count for completeness
winner_stats['non_winners'] = winner_stats['total_participants'] - winner_stats['total_winners']

# Display the results
print("Winner Statistics by Year:")
print("=" * 50)
for year in winner_stats.index:
    if pd.notna(year):  # Skip rows where year is NaN
        year_int = int(year)
        total = winner_stats.loc[year, 'total_participants']
        winners = winner_stats.loc[year, 'total_winners']
        percentage = winner_stats.loc[year, 'winner_percentage']
        
        print(f"{year_int}: {percentage}% were winners ({winners} out of {total} participants)")

print("\nDetailed breakdown:")
print(winner_stats)

# Create a summary visualization
import matplotlib.pyplot as plt

# Filter out NaN years for plotting
clean_stats = winner_stats.dropna()

plt.figure(figsize=(10, 6))
plt.plot(clean_stats.index, clean_stats['winner_percentage'], marker='o', linewidth=2, markersize=8)
plt.title('Winner Percentage by Year', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Winner Percentage (%)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(clean_stats.index)

# Add percentage labels on each point
for year, percentage in zip(clean_stats.index, clean_stats['winner_percentage']):
    plt.annotate(f'{percentage}%', (year, percentage), 
                textcoords="offset points", xytext=(0,10), ha='center')

plt.tight_layout()
plt.show()

# Also create a bar chart showing the trend
plt.figure(figsize=(10, 6))
bars = plt.bar(clean_stats.index, clean_stats['winner_percentage'], 
               color='#A23B72', alpha=0.8, edgecolor='white', linewidth=1)

plt.title('Winner Percentage by Year', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Winner Percentage (%)', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')

# Add percentage labels on top of bars
for bar, percentage in zip(bars, clean_stats['winner_percentage']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
             f'{percentage}%', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```
